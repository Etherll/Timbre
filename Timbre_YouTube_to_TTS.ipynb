| Step | What it does | Time |
|------|--------------|------|
| **1 · Setup** | Installs everything (run once) | ~5–10 min |
| **2 · Input** | Paste a YouTube link or upload a file | ~1 min |
| **3 · Voice sample** | Listen to candidates and pick your speaker | ~1 min |
| **4 · Tweaks** | Optional — safe to skip | — |
| **5 · Run** | Builds and downloads your dataset | ~20–40 min |

Needs a free **T4 GPU**: `Runtime → Change runtime type → T4 GPU → Save`. Then just run the cells top to bottom.

```python
#@title Step 1 · Setup — run once (~5–10 min) { display-mode: "form" }
#@markdown Installs everything and checks the GPU. Run this one cell, then go to Step 2.
import os, subprocess, sys
import shutil as _shutil

# 0) GPU gate — fail fast, before any slow installs.
try:
    _smi = subprocess.run(['nvidia-smi', '--query-gpu=name', '--format=csv,noheader'],
                          capture_output=True, text=True)
    GPU_NAME = (_smi.stdout.strip().splitlines()[0]
                if _smi.returncode == 0 and _smi.stdout.strip() else None)
except FileNotFoundError:
    GPU_NAME = None
if not GPU_NAME:
    raise RuntimeError(
        '❌ No GPU detected. Go to Runtime > Change runtime type, pick "T4 GPU", '
        'click Save, then run this cell again.')
print(f'GPU: {GPU_NAME} ✓')


def _sh(args, label):
    """Run a command quietly; print a friendly line; return CompletedProcess."""
    print(f'{label}…', end=' ', flush=True)
    r = subprocess.run(args, capture_output=True, text=True)
    print('✓' if r.returncode == 0 else '⚠')
    return r


# 1) System audio tools.
_sh(['apt-get', '-qq', 'install', '-y', 'libsndfile1', 'ffmpeg', 'git'], 'Installing audio tools')

# 2) NeMo build prerequisites (must precede NeMo).
_sh([sys.executable, '-m', 'pip', '-q', 'install', 'Cython', 'packaging'], 'Preparing build tools')

# 3) Torch probe — keep Colab's torch when it is new enough + CUDA works.
TORCH_REINSTALLED = False
_keep = False
try:
    import torch as _t
    _ver = tuple(int(x) for x in _t.__version__.split('+')[0].split('.')[:2])
    if _ver >= (2, 7) and _t.cuda.is_available():
        _keep = True
        print(f'Using existing torch {_t.__version__} (CUDA {_t.version.cuda}) ✓')
except Exception:
    _keep = False
if not _keep:
    _sh([sys.executable, '-m', 'pip', '-q', 'install', 'torch>=2.7', 'torchvision', 'torchaudio'],
        'Installing torch')
    TORCH_REINSTALLED = True

# 4) Heavy speech stack (explicit list so no transitive pin swaps torch).
_stack = [
    'nemo_toolkit[asr] @ git+https://github.com/NVIDIA/NeMo.git',
    'audio-separator[gpu]',
    'wespeaker @ git+https://github.com/wenet-e2e/wespeaker.git',
    'fireredvad', 'openai-whisper', 'speechbrain', 'pyannote.core',
    'librosa', 'pyloudnorm', 'transformers', 'onnx', 'onnxruntime',
    'soundfile', 'pyyaml', 'scipy', 'rich', 'ffmpeg-python', 'pandas', 'tqdm',
]
_sh([sys.executable, '-m', 'pip', '-q', 'install'] + _stack, 'Installing Timbre stack (a few minutes)')
_sh([sys.executable, '-m', 'pip', '-q', 'install', '-U', 'yt-dlp'], 'Installing YouTube downloader')
# yt-dlp needs a JS runtime (deno) to solve YouTube's player challenges.
if _shutil.which('deno'):
    print('JS runtime (deno) already installed ✓')
else:
    _sh(['bash', '-c', 'curl -fsSL https://deno.land/install.sh | DENO_INSTALL=/usr/local sh'],
        'Installing JS runtime for YouTube (deno)')

# 5) Clone the repo (idempotent) and enter it.
if not os.path.isdir('/content/Timbre'):
    _sh(['git', 'clone', '-q', 'https://github.com/Etherll/Timbre', '/content/Timbre'], 'Downloading Timbre')
else:
    print('Timbre already downloaded ✓')
os.chdir('/content/Timbre')

# 6) Voice-activity model (REQUIRED — the pipeline refuses to start without it).
print('Downloading voice-activity model…', end=' ', flush=True)
try:
    from huggingface_hub import snapshot_download
    snapshot_download('FireRedTeam/FireRedVAD', local_dir='pretrained_models/FireRedVAD')
    print('✓')
except Exception as _e:
    print(f'⚠ ({type(_e).__name__}: {_e})')

# 7) Preflight gate — import the stack, a CUDA smoke test, and run_timbre.py --help.
print('Checking the install…', end=' ', flush=True)
_failures = []
for _mod in ['torch', 'nemo', 'speechbrain', 'audio_separator',
             'librosa', 'pyloudnorm', 'whisper', 'pandas']:
    try:
        __import__(_mod)
    except Exception as _e:
        _failures.append(f'import {_mod}: {type(_e).__name__}: {_e}')
# WeSpeaker needs Timbre's torchaudio-2.x compat shim (it imports s3prl, which calls the
# removed torchaudio.set_audio_backend). That shim lives in audio_pipeline.py and runs on
# import, exactly as run_timbre.py does it — so verify WeSpeaker via audio_pipeline,
# not a naked 'import wespeaker'.
if '/content/Timbre' not in sys.path:
    sys.path.insert(0, '/content/Timbre')
try:
    import audio_pipeline as _ap
    if not getattr(_ap, 'HAVE_WESPEAKER', False):
        _failures.append('WeSpeaker did not load even with the torchaudio compatibility shim')
except Exception as _e:
    _failures.append(f'load speaker models (audio_pipeline): {type(_e).__name__}: {_e}')
try:
    import torch
    torch.zeros(1).cuda()
except Exception as _e:
    _failures.append(f'GPU check (torch.zeros(1).cuda()): {_e}')
from pathlib import Path as _Path
_vad_dir = _Path('/content/Timbre/pretrained_models/FireRedVAD/VAD')
if not (_vad_dir.is_dir() and any(_vad_dir.iterdir())):
    _failures.append('voice-activity model (FireRedVAD) did not download — '
                     'check your internet connection, then run this cell again')
_help = subprocess.run([sys.executable, 'run_timbre.py', '--help'], capture_output=True, text=True)
if _help.returncode != 0:
    _failures.append('run_timbre.py --help failed:\n' + (_help.stderr or _help.stdout)[:800])
print('done')

if _failures:
    raise RuntimeError(
        '❌ Setup hit a problem:\n  - ' + '\n  - '.join(_failures) +
        '\n\nFix: go to Runtime > Restart session, then run this cell again.')

if TORCH_REINSTALLED:
    print()
    print('⚠  torch was updated — go to Runtime > Restart session, then run this cell once more.')
else:
    print()
    print('✅ Ready — go to Step 2.')
```

```python
#@title Step 2 · Make my dataset { display-mode: "form" }
#@markdown Paste a **YouTube link** (or switch to **Upload a file**) and name the speaker.
#@markdown If YouTube says *"Sign in to confirm you're not a bot"*, tick `youtube_cookies` and upload a
#@markdown `cookies.txt` exported from your browser while logged in to YouTube
#@markdown (browser extension: **Get cookies.txt LOCALLY**).
import os, glob, subprocess

input_source = "YouTube link"  #@param ["YouTube link", "Upload a file"]
youtube_url = "https://www.youtube.com/watch?v=7ARBJQn6QkM"  #@param {type:"string"}
speaker_name = "JensenHuang"  #@param {type:"string"}
youtube_cookies = False  #@param {type:"boolean"}

assert os.path.isdir('/content/Timbre'), 'Run Step 1 (Setup) first — it installs the tools this step needs.'
assert speaker_name.strip(), 'Please enter a speaker_name.'
TARGET_NAME = speaker_name.strip()
INPUT_MEDIA = None


def _dur(path):
    r = subprocess.run(['ffprobe', '-v', 'error', '-show_entries', 'format=duration',
                        '-of', 'default=noprint_wrappers=1:nokey=1', str(path)],
                       capture_output=True, text=True)
    try:
        return float(r.stdout.strip())
    except ValueError:
        return None


if input_source == 'YouTube link':
    assert youtube_url.strip(), 'Paste a YouTube link (or switch input_source to "Upload a file").'
    _cookie_args = []
    if youtube_cookies:
        from google.colab import files
        print('Upload your cookies.txt (exported from your browser while logged in to YouTube):')
        _up = files.upload()
        assert _up, 'No file uploaded.'
        _cookie_args = ['--cookies', os.path.abspath(list(_up.keys())[0])]
    for _old in glob.glob('/content/input.*'):  # clear leftovers from a previous run
        os.remove(_old)
    print('Downloading audio…')
    # bestaudio[ext=webm] = highest-bitrate Opus source -> better TTS dataset bandwidth
    cmd = (['yt-dlp', '-x',
            '-f', 'bestaudio[ext=webm]/bestaudio/best', '--audio-quality', '0',
            '--audio-format', 'wav', '--no-playlist'] + _cookie_args +
           ['-o', '/content/input.%(ext)s', youtube_url])
    _r = subprocess.run(cmd, capture_output=True, text=True)
    if _r.returncode != 0:
        _err = (_r.stderr or _r.stdout or '').strip()
        print('YouTube download failed. Details:')
        print(_err[-600:])
        print()
        if 'not a bot' in _err or 'cookies' in _err:
            raise SystemExit(
                'YouTube is blocking Colab\'s servers for this video. Two easy fixes:\n'
                '  1. Tick "youtube_cookies" above and re-run this cell — you will be asked to upload\n'
                '     a cookies.txt (export it with the "Get cookies.txt LOCALLY" browser extension\n'
                '     while logged in to YouTube), or\n'
                '  2. Switch input_source to "Upload a file" and upload the video/audio yourself.')
        raise SystemExit('Download failed — check the link, or switch input_source to "Upload a file".')
    _outs = glob.glob('/content/input.*')
    _wavs = [p for p in _outs if p.lower().endswith('.wav')]
    INPUT_MEDIA = _wavs[0] if _wavs else (max(_outs, key=os.path.getmtime) if _outs else '/content/input.wav')
else:
    from google.colab import files
    print('Choose a file to upload (mp4 / wav / mp3 / m4a …)')
    _up = files.upload()
    assert _up, 'No file uploaded.'
    INPUT_MEDIA = os.path.abspath(list(_up.keys())[0])

INPUT_MEDIA = os.path.abspath(INPUT_MEDIA)
assert os.path.isfile(INPUT_MEDIA), (
    f'No input file produced (looked for {INPUT_MEDIA}). Try the "Upload a file" option.')
_d = _dur(INPUT_MEDIA)
print(f'✓ Got {os.path.basename(INPUT_MEDIA)}' + (f'  ({_d/60:.1f} min)' if _d else '')
      + f' for speaker "{TARGET_NAME}".')
print('Go to Step 3 to pick the voice sample.')
```

```python
#@title Step 3 · Pick the voice sample — listen & choose { display-mode: "form" }
#@markdown This finds clean-speech moments in your audio and shows them 6 at a time. **Listen, then
#@markdown click the button under the sample that is only your speaker** (no music, no other voices).
#@markdown Wrong person in the first 6? Click **Show 6 more samples**.
#@markdown Or tick the box to upload your own ~10 s clip instead.
import os, sys, glob, subprocess

upload_my_own_clip = False  #@param {type:"boolean"}

assert os.path.isdir('/content/Timbre'), 'Run Step 1 (Setup) first.'
os.chdir('/content/Timbre')
INPUT_MEDIA = globals().get('INPUT_MEDIA')
assert INPUT_MEDIA, 'No input yet — run Step 2 first.'

REFERENCE_WAV = None


def _dur(path):
    r = subprocess.run(['ffprobe', '-v', 'error', '-show_entries', 'format=duration',
                        '-of', 'default=noprint_wrappers=1:nokey=1', str(path)],
                       capture_output=True, text=True)
    try:
        return float(r.stdout.strip())
    except ValueError:
        return 0.0


if upload_my_own_clip:
    from google.colab import files
    print('Upload a clean ~10 s clip of just your target speaker:')
    _up = files.upload()
    assert _up, 'No file uploaded.'
    _raw = os.path.abspath(list(_up.keys())[0])
    subprocess.run(['ffmpeg', '-hide_banner', '-loglevel', 'error', '-y', '-i', _raw,
                    '-vn', '-ac', '1', '-ar', '16000', '/content/ref_uploaded_16k.wav'], check=True)
    REFERENCE_WAV = '/content/ref_uploaded_16k.wav'
    print(f'✓ Will use your uploaded clip ({_dur(REFERENCE_WAV):.1f}s). Go to Step 5 (or Step 4 to tweak).')
else:
    for _old in glob.glob('/content/ref_clips/ref_*.wav'):  # clear picks from a previous video
        os.remove(_old)
    print('Looking for clean-speech moments…')
    subprocess.run([sys.executable, 'extract_reference.py', '-i', INPUT_MEDIA,
                    '-o', '/content/ref_clips', '--min-clip', '3', '--max-clip', '15',
                    '--longest-first', '--limit', '30'], capture_output=True, text=True)
    _cands = sorted(glob.glob('/content/ref_clips/ref_*.wav'))  # ref_000 = longest speech moment
    if not _cands:
        raise RuntimeError(
            'Could not find clean speech in this audio. Tick "upload_my_own_clip" above, '
            're-run this cell, and upload a ~10 s clip of your speaker.')

    import ipywidgets as widgets
    from IPython.display import Audio, display

    REFERENCE_WAV = os.path.abspath(_cands[0])  # default — the buttons below override it

    print(f'Found {len(_cands)} samples. Sample 1 is selected by default — click a button to change it.')
    print('None of them the right person? Tick "upload_my_own_clip" above and re-run this cell.')
    print()

    _PAGE = 6
    _shown = 0
    _buttons = {}
    _area = widgets.Output()

    def _select(idx):
        def _cb(_btn):
            globals()['REFERENCE_WAV'] = os.path.abspath(_cands[idx])
            for _j, _b in _buttons.items():
                _b.button_style = 'success' if _j == idx else ''
                _b.description = (f'✓ Using sample {_j + 1}' if _j == idx else f'Use sample {_j + 1}')
        return _cb

    def _show_more(_btn=None):
        global _shown
        _end = min(_shown + _PAGE, len(_cands))
        with _area:
            for _i in range(_shown, _end):
                display(widgets.HTML(f'<b>Sample {_i + 1}</b> &nbsp;·&nbsp; {_dur(_cands[_i]):.1f}s'))
                display(Audio(_cands[_i]))
                _b = widgets.Button(
                    description=('✓ Using sample 1' if _i == 0 else f'Use sample {_i + 1}'),
                    button_style=('success' if _i == 0 else ''),
                    layout=widgets.Layout(width='auto'))
                _b.on_click(_select(_i))
                _buttons[_i] = _b
                display(_b)
        _shown = _end
        _left = len(_cands) - _shown
        _more.disabled = _left <= 0
        _more.description = (f'Show {min(_PAGE, _left)} more samples' if _left > 0
                             else 'No more samples in this audio')

    _more = widgets.Button(description='Show 6 more samples', icon='plus',
                           layout=widgets.Layout(width='auto'))
    _more.on_click(_show_more)
    display(_area)
    _show_more()
    display(_more)
    print()
    print('Happy with your pick? Go to Step 5 to run (or Step 4 to tweak settings).')
```

```python
#@title Step 4 · Optional tweaks — safe to skip { display-mode: "form" }
#@markdown **Skip this cell unless you want to change something** — good defaults are used otherwise.

language = "en"  #@param ["en", "es", "fr", "de", "it", "pt", "auto"] {allow-input: true}
#@markdown ↑ Transcription language ("auto" = detect).
match_strictness = 0.7  #@param {type:"slider", min:0.3, max:0.9, step:0.05}
#@markdown ↑ Higher = stricter voice match (fewer, purer clips). Lower = keeps more audio.
asr_engine = "nemotron"  #@param ["nemotron", "whisper"]
#@markdown ↑ Speech-to-text engine for the transcripts.
quick_test = False  #@param {type:"boolean"}
#@markdown ↑ Fast ~60 s sanity run (still downloads models) — check the chain before a full run.

print('✓ Settings saved — go to Step 5.')
```

```python
#@title Step 5 · Run & download — click and wait { display-mode: "form" }
#@markdown Runs Timbre with the voice sample you picked in Step 3, then downloads your dataset.
#@markdown Takes ~20–40 min; long quiet stretches while models download are normal.
import os, sys, glob, random, subprocess, shlex, time
from pathlib import Path

assert os.path.isdir('/content/Timbre'), 'Run Step 1 (Setup) first.'
os.chdir('/content/Timbre')
if '/content/Timbre' not in sys.path:
    sys.path.insert(0, '/content/Timbre')
import torch

INPUT_MEDIA = globals().get('INPUT_MEDIA')
TARGET_NAME = globals().get('TARGET_NAME')
assert INPUT_MEDIA and TARGET_NAME, 'No input yet — run Step 2 first.'

# ---- Settings from Steps 3–4, with safe defaults so skipping them never errors. ----
language = globals().get('language', 'en')
match_strictness = globals().get('match_strictness', 0.7)
quick_test = globals().get('quick_test', False)
asr_engine = globals().get('asr_engine', 'nemotron')
REFERENCE_WAV = globals().get('REFERENCE_WAV')


def _dur(path):
    r = subprocess.run(['ffprobe', '-v', 'error', '-show_entries', 'format=duration',
                        '-of', 'default=noprint_wrappers=1:nokey=1', str(path)],
                       capture_output=True, text=True)
    try:
        return float(r.stdout.strip())
    except ValueError:
        return 0.0


# ---- (a) Voice sample: from Step 3, or a silent auto-pick if Step 3 was skipped. ----
if not REFERENCE_WAV:
    for _old in glob.glob('/content/ref_clips/ref_*.wav'):
        os.remove(_old)
    subprocess.run([sys.executable, 'extract_reference.py', '-i', INPUT_MEDIA,
                    '-o', '/content/ref_clips', '--min-clip', '3', '--max-clip', '15',
                    '--longest-first', '--limit', '1'], capture_output=True, text=True)
    _cands = sorted(glob.glob('/content/ref_clips/ref_*.wav'))
    if not _cands:
        raise RuntimeError(
            'Could not auto-pick a voice sample from this audio. Go to Step 3, tick '
            '"upload_my_own_clip", upload a clean ~10 s clip of your speaker, then re-run this cell.')
    REFERENCE_WAV = os.path.abspath(_cands[0])
    print(f'Voice sample: auto-picked ({_dur(REFERENCE_WAV):.1f}s). '
          'Tip: run Step 3 to listen and choose it yourself.')
else:
    REFERENCE_WAV = os.path.abspath(REFERENCE_WAV)
    print(f'Voice sample: your pick from Step 3 ({_dur(REFERENCE_WAV):.1f}s).')

# ---- (b) Voice-activity model is required — self-heal if it is missing. ----
_vad_dir = Path('/content/Timbre/pretrained_models/FireRedVAD/VAD')
if not (_vad_dir.is_dir() and any(_vad_dir.iterdir())):
    print('Voice-activity model missing — downloading it now…')
    from huggingface_hub import snapshot_download
    snapshot_download('FireRedTeam/FireRedVAD', local_dir='/content/Timbre/pretrained_models/FireRedVAD')
    print('Voice-activity model ready ✓')

# ---- (c) GPU autotune: small / pre-Ampere GPU (e.g. the free T4) -> low-vram + fp16. ----
_gpu = torch.cuda.get_device_name(0) if torch.cuda.is_available() else ''
_cap = torch.cuda.get_device_capability() if torch.cuda.is_available() else (0, 0)
_is_small = ('T4' in _gpu) or (_cap < (8, 0))
_eff_precision = 'fp16' if _is_small else 'fp32'

# ---- (d) Build and run the command. ----
cmd = [sys.executable, 'run_timbre.py',
       '-i', INPUT_MEDIA, '-r', REFERENCE_WAV, '-n', TARGET_NAME,
       '--output-base-dir', '/content/output_runs',
       '--verification-threshold', str(match_strictness),
       '--language', language,
       '--asr-backend', asr_engine,
       '--asr-precision', _eff_precision]
if _is_small:
    cmd += ['--low-vram']
if quick_test:
    cmd += ['--dry-run']

print()
print('Working… this takes ~20–40 min. Long quiet stretches while models download are normal — don’t stop the cell.')
print('Command: ' + ' '.join(shlex.quote(c) for c in cmd))
print('=' * 60)
_t0 = time.monotonic()
proc = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1)
for _line in proc.stdout:
    print(_line, end='')
_rc = proc.wait()
_mins = (time.monotonic() - _t0) / 60
print('=' * 60)

# ---- (e) Locate the dataset, then summarize + preview + download (or explain what to do). ----
from common import safe_filename
import pandas as pd
from IPython.display import Audio, display

run_dir = Path('/content/output_runs') / f'{safe_filename(TARGET_NAME)}_{Path(INPUT_MEDIA).stem}_extracted'
dataset_dir = run_dir / 'dataset'
meta_path = dataset_dir / 'metadata.csv'

_df, _rows = None, 0
if meta_path.is_file():
    try:
        _df = pd.read_csv(meta_path, sep='|', header=None,
                          names=['id', 'transcript', 'normalized'], engine='python')
        _rows = len(_df)
    except Exception:
        _df, _rows = None, 0

if _rows == 0:
    if _rc != 0:
        print('❌ The run stopped with an error — scroll up for the last lines of the log.')
        print('Fix: Runtime > Restart session, re-run Step 1, then Steps 2–5.')
        print('Still failing? Open Step 4 and tick "quick_test" to sanity-check the chain first.')
    else:
        print('No clips were produced.')
        print('The voice sample was probably the wrong person, or the audio had a lot of')
        print('overlapping speech. Go back to Step 3, pick a different sample (or upload')
        print('your own clean ~10 s clip), then re-run this cell.')
else:
    import soundfile as sf
    _wavs = glob.glob(str(dataset_dir / 'wavs' / '*' / '*.wav'))
    _total_s = 0.0
    for _w in _wavs:
        try:
            _i = sf.info(_w)
            _total_s += _i.frames / float(_i.samplerate)
        except Exception:
            pass
    _train = dataset_dir / 'train.csv'
    _eval = dataset_dir / 'eval.csv'
    _train_n = sum(1 for _ in open(_train)) if _train.is_file() else 0
    _eval_n = sum(1 for _ in open(_eval)) if _eval.is_file() else 0
    if _rc != 0:
        print('⚠ The run reported an error but still produced a dataset — check the log above.')
    print(f'✅ Done in {_mins:.0f} min — {_rows} clips ({_train_n} train / {_eval_n} eval), '
          f'{_total_s/60:.1f} min of audio.')
    print()
    print('Sample of metadata.csv:')
    print(_df.head(5).to_string(index=False))
    if _wavs:
        _pick = random.choice(_wavs)
        _match = _df[_df['id'] == Path(_pick).stem]
        print()
        print('Listen to one clip:')
        if len(_match):
            print('  “' + str(_match.iloc[0]['transcript']) + '”')
        display(Audio(_pick))
    # Zip + download.
    import shutil
    _zip = shutil.make_archive('/content/Timbre_dataset', 'zip', root_dir=str(run_dir), base_dir='dataset')
    print()
    print(f'Downloading {os.path.basename(_zip)}…')
    from google.colab import files
    files.download(_zip)
```

## What you got

`Timbre_dataset.zip` is a standard **LJSpeech** dataset: `dataset/metadata.csv`
(`id|transcript|normalized_transcript`) and `dataset/wavs/<speaker>/<id>.wav`, plus a `train.csv` /
`eval.csv` split. Point any LJSpeech-style TTS trainer (Coqui TTS, VITS, Piper, an XTTS fine-tune)
at the `dataset/` folder. Clips are word-safe (no mid-word cuts) and loudness-normalized.

## If something went wrong

- **No clips produced** — the voice sample was probably the wrong person. Go back to Step 3,
  pick a different sample (or upload your own clean ~10 s clip), and re-run Step 5.
- **YouTube says "Sign in to confirm you're not a bot"** — YouTube blocks Colab's servers for some
  videos. In Step 2 tick `youtube_cookies` and upload a `cookies.txt` exported from your browser
  (extension: **Get cookies.txt LOCALLY**, while logged in to YouTube) — or switch to **Upload a file**.
- **Setup or run errors** — `Runtime → Restart session`, then re-run the steps from the top.

More options and the full field reference: the
[Timbre README](https://github.com/Etherll/Timbre#readme).
